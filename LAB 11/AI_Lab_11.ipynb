```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler, LabelEncoder

data = {
    'vehicle_serial_no':,
    'mileage':,
    'fuel_efficiency':,
    'maintenance_cost':,
    'vehicle_type': ['SUV', 'Sedan', 'Truck', 'Hatchback', 'Sedan', 'Truck', 'SUV', 'Truck', 'Hatchback', 'SUV']
}
df_vehicle = pd.DataFrame(data)

# Encode categorical vehicle_type
le = LabelEncoder()
df_vehicle['vehicle_type_num'] = le.fit_transform(df_vehicle['vehicle_type'])

X_vehicle = df_vehicle[['vehicle_serial_no', 'mileage', 'fuel_efficiency', 'maintenance_cost', 'vehicle_type_num']]

# 1. Without Scaling
km_veh_no_scale = KMeans(n_clusters=3, random_state=42).fit(X_vehicle)
df_vehicle['cluster_no_scale'] = km_veh_no_scale.labels_

# 2. With Scaling (Except vehicle_type)
scaler_v = StandardScaler()
scale_cols = ['vehicle_serial_no', 'mileage', 'fuel_efficiency', 'maintenance_cost']
X_v_scaled = X_vehicle.copy()
X_v_scaled[scale_cols] = scaler_v.fit_transform(X_vehicle[scale_cols])

km_veh_scaled = KMeans(n_clusters=3, random_state=42).fit(X_v_scaled)
df_vehicle['cluster_scaled'] = km_veh_scaled.labels_

# Insight: 'mileage' numbers are huge compared to 'fuel_efficiency'.
# Without scaling, the model only groups by mileage.
```
